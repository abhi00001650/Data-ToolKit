```python
# details in the chart

plt.plot(x,y , color = 'red')
plt.title('line chart to understand syntax')
plt.xlabel('number of day')
plt.ylabel('price of stock')
plt.show()
```

```python
# details in the chart 
# to add color and meetup point as a point

plt.plot(x,y , color = 'red' , marker = 'o')
plt.title('line chart to understand syntax')
plt.xlabel('number of day')
plt.ylabel('price of stock')
plt.show()
```

```python
# details in the chart
# to add lin style and inc. line width
plt.plot(x,y , color = 'red' , marker = 'o' , linestyle = '--' , linewidth = 3 )
plt.title('line chart to understand syntax')
plt.xlabel('number of day')
plt.ylabel('price of stock')
plt.show()
```

```python
#inc. marker size
# to add grids
plt.plot(x,y , color = 'red' , marker = 'o' , linestyle = '--' , linewidth = 3 , markersize = 10 )
plt.title('line chart to understand syntax')
plt.xlabel('number of day')
plt.ylabel('price of stock')
plt.grid()
plt.show()
```

```python
x = np.linspace(0,10,100)
```

```python
#different types of line shapes
plt.plot(x , np.sin(x) , linestyle = '--')
plt.plot(x , np.cos(x) , linestyle = ':')
plt.plot(x , np.tan(x) , linestyle = '-.')
```

```python
plt.figure() # initializing a page to make plot

plt.subplot(1,2,1) # 3 parameters 1- row (plot) , 2- columns (subplots) , 3- (sin , cos)
plt.plot(x , np.sin(x))

plt.subplot(1,2,2) 
plt.plot(x , np.cos(x))


plt.show
```

```python
# adding more colums for figurs by just changing dimensions(-,-,-)
plt.figure() # initializing a page to make plot

plt.subplot(3,1,1) # 3 parameters 1- row (plot) , 2- columns (subplots) , 3- (sin , cos)
plt.plot(x , np.sin(x))

plt.subplot(3,1,2) 
plt.plot(x , np.cos(x))

plt.subplot(3,1,3) 
plt.plot(x , np.tan(x))


plt.show
```

```python
# 4 plots
plt.figure() # initializing a page to make plot

plt.subplot(2,2,1) # 3 parameters 1- row (plot) , 2- columns (subplots) , 3- (sin , cos)
plt.plot(x , np.sin(x) , linestyle = '-' , color = 'r')

plt.subplot(2,2,2) 
plt.plot(x , np.cos(x) , linestyle = '--' , color = 'b')

plt.subplot(2,2,3) 
plt.plot(x , np.tan(x) , linestyle = '-.' , color = 'g')

plt.subplot(2,2,4) 
plt.plot(x , np.sin(x+1) , linestyle = ':' , color = 'y') # NOTE - SIN(x+1) as cot , cosec ,sec is not defined in np


plt.show
```

```python
# 6 plots >> 2 rows , 3 columns
plt.figure() # initializing a page to make plot

plt.subplot(2,3,1) # 3 parameters 1- row (plot) , 2- columns (subplots) , 3- (sin , cos)
plt.plot(x , np.sin(x) , linestyle = '-' , color = 'r')

plt.subplot(2,3,2) 
plt.plot(x , np.cos(x) , linestyle = '--' , color = 'b')

plt.subplot(2,3,3) 
plt.plot(x , np.tan(x) , linestyle = '-.' , color = 'g')

plt.subplot(2,3,4) 
plt.plot(x , np.sin(x+1) , linestyle = ':' , color = 'y') # NOTE - SIN(x+1) as cot , cosec ,sec is not defined in np

plt.subplot(2,3,5) 
plt.plot(x , np.cos(x+1) , linestyle = '-' , color = 'k')

plt.subplot(2,3,6) 
plt.plot(x , np.tan(x+1) , linestyle = '-.' , color = 'c')

plt.show
```

```python
#scatterplot >> to understand the relationship between two features
x = np.random.rand(50)
y = np.random.rand(50)


plt.scatter(range(50), x, color='green', marker = 'X' , label='x values')
plt.scatter(range(50), y, color='red', label='y values')
plt.xlabel('Index')
plt.ylabel('Value')
plt.title('Scatter Plot of x and y')
plt.legend()
plt.show()
```

```python
# pie chart >> used to understand the percent distribution of data

langs = ['c','python','java']
student = [20,100,40]

plt.pie(student , labels = langs , autopct = '%1.1f%%')
plt.title('Pie_chart')
plt.show()
```

```python
#FOCUS ON 1 MORE SEGMENT
explode = (0.0 , 0.09 , 0.09)
langs = ['c','python','java']
student = [20,100,40]

plt.pie(student , labels = langs , autopct = '%1.1f%%' , explode = explode , shadow = True )
plt.title('Pie_chart')
plt.show()
```

```python
#area plot >> variables changed over time

#An area plot in Matplotlib is a type of plot that displays quantitative data visually, showing how one or more variables change over a continuous range (usually time). It’s similar to a line plot, but the area under the line is filled with color, emphasizing the magnitude of values.

```

```python
years = [2021,2018,2022,2023,2017]
category1 = [20,25,30,28,35]
category2 = [10,15,20,18,25]
category3 = [5,8,10,9,15]

plt.stackplot(years , category1 , category2 , category3 , labels = ['category1' , 'category2' , 'category3'] ,colors = ['r','g')
```
